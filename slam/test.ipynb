# Suggested Learning Milestones

- [ ] **1. Read LiDAR data**
  - Subscribe to `/scan`
  - Parse `sensor_msgs/msg/LaserScan`
  - Visualize scan points in RViz

- [ ] **2. Read robot odometry**
  - Subscribe to `/odom`
  - Parse `nav_msgs/msg/Odometry`
  - Understand the robot pose (`x`, `y`, `θ`)

- [ ] **3. Transform laser scans into the world frame**
  - Convert scan points from the robot frame to the map frame
  - Verify the transformed points in RViz

- [ ] **4. Create an occupancy grid**
  - Implement a 2D grid map
  - Choose a map resolution and dimensions
  - Publish a `nav_msgs/msg/OccupancyGrid`

- [ ] **5. Implement Bresenham ray tracing**
  - Mark cells along each laser beam as free
  - Mark obstacle endpoints as occupied
  - Continuously update the occupancy grid

- [ ] **6. Build a mapping-only SLAM node**
  - Assume odometry is perfect
  - Generate a consistent occupancy map while the robot explores

- [ ] **7. Implement scan matching**
  - Compare incoming scans to previous scans or the map
  - Estimate pose corrections instead of relying solely on odometry
  - Explore algorithms such as ICP or Correlative Scan Matching

- [ ] **8. Replace odometry-only localization**
  - Fuse odometry with scan matching
  - Reduce accumulated drift
  - Publish an improved `map → odom` transform

- [ ] **9. Build a pose graph**
  - Store robot poses as graph nodes
  - Add odometry and scan-matching constraints as graph edges

- [ ] **10. Detect loop closures**
  - Recognize when the robot revisits a previously mapped area
  - Add loop closure constraints to the pose graph

- [ ] **11. Optimize the pose graph**
  - Use a graph optimization library (e.g., g2o, GTSAM, or Ceres)
  - Minimize accumulated localization error
  - Produce a globally consistent map

- [ ] **12. Evaluate and compare**
  - Test in increasingly complex Gazebo worlds
  - Record and replay datasets with `ros2 bag`
  - Compare your results against SLAM Toolbox or Cartographer
  - Measure map quality, localization accuracy, and runtime performance
